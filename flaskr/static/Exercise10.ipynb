```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('sample2.csv')

df = pd.DataFrame(data)

print(df.columns)
```

```python
df_labels = df['issue_label'].value_counts()
labels = np.array(df_labels.index)
count = np.array(df_labels)
print(f'{labels}{count}')

fig, ax = plt.subplots()
bar_colors = ['tab:red', 'tab:blue', 'tab:orange']

ax.bar(labels, count, color=bar_colors)
ax.set_title('Labels frequency')
ax.set_xlabel('Labels')

plt.show()
```

```python
df_authors = df['issue_author_association'].value_counts()

authors = np.array(df_authors.index)
count = np.array(df_authors)
print(df_authors)

#Graph
fig, ax = plt.subplots()

ax.bar(authors, count)
ax.set_title('Labels frequency')
ax.set_xlabel('Labels')

plt.show()
```

```python

df['created_at'] = pd.to_datetime(df['issue_created_at']).dt.strftime('%Y-%m-%d')

df_orderby_date = df.sort_values(by='created_at') 
#print(df_orderby_date['created_at'])
dates_dis = df['created_at'].value_counts().sort_index()
#print(dates_dis)

date = np.array(dates_dis.index)
count = np.array(dates_dis)

#Graph
fig, ax = plt.subplots()

ax.bar(date, count, color = 'tab:purple')
ax.set_title('Distribution')
fig.set_size_inches(50,50)
#ax.set_xlabel('Labels')

plt.show()
```
