```python
###Libraries
```

```python
import numpy as np
import pandas as pd
import nltk
```

```python
###Read csv
```

```python
data = pd.read_csv('sample1.csv')
text_data = data[['issue_title','issue_body']]
print(text_data)
```

```python
###Noise reduction
```

```python
import string
string.punctuation

text_data.loc[:,'issue_title'] = text_data['issue_title'].str.lower()
text_data.loc[:,'issue_body'] = text_data['issue_body'].str.lower()

def remove_punctuation(text):
    if not isinstance(text, str):
        text = ""
    punctuationFree = "".join([i for i in text if i not in string.punctuation])
    return punctuationFree

text_data.loc[:,'issue_title'] = text_data['issue_title'].apply(lambda x:remove_punctuation(x))
text_data.loc[:,'issue_body'] = text_data['issue_body'].apply(lambda x:remove_punctuation(x))
text_data.head()
```

```python
###Tokenization
```

```python
#nltk.download('punkt')

def tokenization(text):
    tokens = nltk.word_tokenize(text)
    repr(tokens)
    return tokens

text_data.loc[:,'token_title'] = text_data['issue_title'].apply(lambda x: tokenization(x))
text_data.loc[:,'token_body'] = text_data['issue_body'].apply(lambda x: tokenization(x))
text_data.head()
```

```python
###Stop word removal
```

```python
#nltk.download('stopwords')
from nltk.corpus import stopwords
def remove_stop_words(text):
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in text if word not in stop_words]
    return tokens

text_data.loc[:,'token_title'] = text_data['token_title'].apply(lambda x:remove_stop_words(x))
text_data.loc[:,'token_body'] = text_data['token_body'].apply(lambda x:remove_stop_words(x))
text_data.head()
```

```python
###Stemming
```

```python
from nltk.stem.porter import PorterStemmer
porter = PorterStemmer()

def stemming(text):
    stem_text = [porter.stem(word) for word in text]
    return stem_text
text_data.loc[:,'token_title'] = text_data['token_title'].apply(lambda x:stemming(x))
text_data.loc[:,'token_body'] = text_data['token_body'].apply(lambda x:stemming(x))
text_data.head()
```

```python
###Lemmatization
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemma = WordNetLemmatizer()

def lemmatization(text):
    tokens = [lemma.lemmatize(word) for word in text]
    return tokens

text_data.loc[:,'token_title'] = text_data['token_title'].apply(lambda x:lemmatization(x))
text_data.loc[:,'token_body'] = text_data['token_body'].apply(lambda x:lemmatization(x))
text_data.head()
```
