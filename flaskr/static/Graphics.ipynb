```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# CSV
df = pd.read_csv('metrics.csv')

# Seaborn
sns.set(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Compare models RF', fontsize=16)

# Accuracy (Validation)
sns.barplot(ax=axes[0, 0], x='name', y='accuracy_val', data=df, palette='viridis')
axes[0, 0].set_title('Accuracy (Validation)')
axes[0, 0].set_ylabel('Accuracy')
axes[0, 0].set_xticklabels(axes[0, 0].get_xticklabels(), rotation=45, ha='right')

# Accuracy (Test)
sns.barplot(ax=axes[0, 1], x='name', y='accuracy', data=df, palette='viridis')
axes[0, 1].set_title('Accuracy (Test)')
axes[0, 1].set_ylabel('Accuracy')
axes[0, 1].set_xticklabels(axes[0, 1].get_xticklabels(), rotation=45, ha='right')

# F1 Score (Validation)
sns.barplot(ax=axes[1, 0], x='name', y='f1_val', data=df, palette='viridis')
axes[1, 0].set_title('F1 Score (Validation)')
axes[1, 0].set_ylabel('F1 Score')
axes[1, 0].set_xticklabels(axes[1, 0].get_xticklabels(), rotation=45, ha='right')

# F1 Score (Test)
sns.barplot(ax=axes[1, 1], x='name', y='f1', data=df, palette='viridis')
axes[1, 1].set_title('F1 Score (Test)')
axes[1, 1].set_ylabel('F1 Score')
axes[1, 1].set_xticklabels(axes[1, 1].get_xticklabels(), rotation=45, ha='right')


plt.tight_layout(rect=[0, 0, 1, 0.96])

# Show
plt.show()
```

```python
import numpy as np
fig, axs = plt.subplots(nrows=1, ncols=1, figsize=(6, 6), sharey=True)

labels = ['Accuracy', 'Accuracy test','F1 Score','F1 Score test']
df_val = df[['accuracy_val','accuracy','f1_val','f1']]


medianprops = dict(linestyle='--', linewidth=2.5, color='firebrick') #Median
meanlineprops = dict(linestyle='--', linewidth=2.5, color='purple') #Mean

vals, names, xs = [],[],[]

for i, col in enumerate(df_val.columns):
    vals.append(df[col].values)
    names.append(col)
    xs.append(np.random.normal(i + 1, 0.04, df[col].values.shape[0]))

axs.boxplot(df_val, tick_labels = labels, meanline=True, showmeans=True, medianprops=medianprops, meanprops =meanlineprops)

palette = ['r', 'r', 'b', 'b']

for x, val, c in zip(xs, vals, palette):
   axs.scatter(x, val, alpha=0.4, color=c)
    
axs.set_title('Accuracy and F1 Score')
axs.legend()

plt.show()
```
